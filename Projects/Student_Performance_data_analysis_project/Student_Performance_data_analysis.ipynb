# Check and correct invalid exam score

```python
        #that exam scoee col contain one marks 101 which is invalid,it must be equla to or less than 100
        #let see how many values are greater then 100

print(df_cleaned[df_cleaned["Exam_Score"] > 100]) #it gives just one row where score is 101

df_cleaned["Exam_Score"] = df_cleaned["Exam_Score"].replace(101, 100) #it replaces 101 to 100

print(df_cleaned["Exam_Score"].max()) #max value of exam_score col is 100 now
```

# Final cleaning verification

```python
                   #our 3 major problems was these now solved
#Verify that missing values and duplicates are handled and that the maximum exam score is 100.

print(df_cleaned.isnull().sum().sum())

print(df_cleaned.duplicated().sum())

print(df_cleaned["Exam_Score"].max())
```

```python
                    #lets use numpy here for statistical calculation for dataset
#Convert exam scores into NumPy array

exam_scores=df_cleaned["Exam_Score"].to_numpy()
print(exam_scores)

print("Mean:",np.mean(exam_scores))

print("Median:",np.median(exam_scores))

print("Minimum:",np.min(exam_scores))

print("Maximum:",np.max(exam_scores))

print("Standard Deviation:",np.std(exam_scores))

print("Variance:",np.var(exam_scores))
```

# Exploratory Data Analysis (EDA)

```python
# EDA

#here we check relatonships between columns,in our dataset main purpose is exam_score now we will see which col is affecting and influencing exam_score from that we will see relatonship
#for relationships between columns we use correlation which is statistical measure which shows direction and relation between two variables
#correlation value lies between -1 to 1
#+1: A perfect positive linear relationship.(mean one varaible increases other also increases)
#0: No linear relationship.(mean one variable does not affect other)
#-1: A perfect negative linear relationship.(opposite of linear relationship,one var is increase other decrease)


## Now that we have successfully learned and applied Pandas and NumPy for data inspection, cleaning, and statistical calculations, we are moving on to Exploratory Data Analysis (EDA) to discover patterns, trends, distributions, and relationships in our dataset.
        #EDA is about What can I discover from my data?
```

# Numerical col rlationship

# hours_studied and exam_score

```python
#first we will get two columns from dataframe and then apply correlation on both
                            #Method 1

                            #check which columns affecting exam_score

                            #Numericals columns relationship

                         

study_exam = df_cleaned[["Hours_Studied", "Exam_Score"]]  #two variables

correlation = study_exam["Hours_Studied"].corr(study_exam["Exam_Score"]) 

print(correlation)
                            #Other Method Easy

"""
correlation = df_cleaned["Hours_Studied"].corr(df_cleaned["Exam_Score"])

print(correlation)

"""

sns.scatterplot(
    data=study_exam,
    x="Hours_Studied",
    y="Exam_Score"
)

plt.title("Study Hours vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()
```

# Attendance vs Exam Score

```python
# EDA Finding:
# Attendance shows a positive relationship with Exam_Score.
# Students with higher attendance tend to have higher exam scores.
# However, students with similar attendance can have different exam scores,
# indicating that other factors may also be associated with performance.

correlation = df_cleaned["Attendance"].corr(df_cleaned["Exam_Score"])

print("Correlation:", correlation)

sns.scatterplot(data=df_cleaned, x="Attendance", y="Exam_Score")
plt.title("Attendance vs Exam Score")
plt.show()
```

# Correlation Heatmap

#Purpose: Visualize correlations between all numerical columns together and identify positive and negative linear relationships.


```python
correlation = df_cleaned.corr(numeric_only=True)

#annot=True adds number in eac box to see correlation if annot is False then heatmap boxes will be blur
#fmt concepts handles round off( how many values after fraction or points)
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```

# Key Insights

1. Most students' exam scores were concentrated around 60–70.
2. Hours studied showed a moderate positive relationship with exam scores.

4. Attendance showed an upward visual trend with exam scores.
5. Average exam scores across different categorical groups showed relatively small differences.

These observations describe relationships and patterns in the dataset, not necessarily cause-and-effect relationships.
