# complent rule

```python

"""
COMPLEMENT RULE OF PROBABILITY

Definition:
The Complement Rule is used to find the probability
that an event does NOT happen.

The complement of an event means that the event does not occur.

Formula:

P(A') = 1 - P(A)

Where:
P(A') = Probability that event A does not happen.
P(A) = Probability that event A happens.
1 = Total probability (100%).

Example:
If the probability of getting Heads is 0.5,
then the probability of NOT getting Heads is:

P(A') = 1 - 0.5
P(A') = 0.5
"""
```

# question

```python

"""
QUESTION 1:

The probability that a student passes an exam is 0.80.
Find the probability that the student does NOT pass.

SOLUTION:

Given:
P(A) = 0.80

Required:
P(A') = ?

Apply complement rule:
P(A') = 1 - P(A)
"""

# Identify event
# A = student passes the exam
# A' = student does not pass the exam

# Given probability
p_a = 0.80

# Apply complement rule
p_not_a = 1 - p_a

print("Final answer is:", p_not_a)
print("Percentage:", p_not_a * 100, "%")
```

```python

"""
QUESTION 2:

A bag contains 5 red balls and 3 blue balls.
One ball is selected randomly.
Find the probability of NOT selecting a red ball.

SOLUTION:

Total balls = 8
Red balls = 5

P(A) = Probability of selecting a red ball
P(A') = Probability of NOT selecting a red ball

P(A) = 5/8

Apply complement rule:
P(A') = 1 - P(A)
"""

# Identify event
# A = selecting a red ball
# A' = not selecting a red ball

# Given probability
p_a = 5 / 8

# Apply complement rule
p_not_a = 1 - p_a

print("Final answer is:", p_not_a)
print("Percentage:", p_not_a * 100, "%")
```
