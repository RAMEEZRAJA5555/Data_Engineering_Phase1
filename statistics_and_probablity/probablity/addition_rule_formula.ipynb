# Addition rule of probablity

```python

"""
ADDITION RULE OF PROBABILITY

Definition:

The Addition Rule of Probability is used to calculate the probability
that at least one of two events will happen.

In simple words:

It helps us find the probability of event A OR event B happening.

Formula:

P(A or B) = P(A) + P(B) - P(A and B)

OR

P(A ∪ B) = P(A) + P(B) - P(A ∩ B)


EXPLANATION OF FORMULA:

P(A):
Probability of event A happening.

P(B):
Probability of event B happening.

P(A and B):
Probability that both events A and B happen together.

P(A or B):
Probability that event A or event B or both happen.

WHY DO WE SUBTRACT P(A and B)?

When we add P(A) and P(B), the common part where both events happen
is counted twice.

Therefore, we subtract P(A and B) once to avoid double counting.


IMPORTANT TERMS:

1. Mutually Exclusive Events:

Two events are mutually exclusive when they cannot happen at the
same time.

For example:
Getting a 2 and getting a 5 in one roll of a die.

Formula:

P(A or B) = P(A) + P(B)

Because:

P(A and B) = 0


2. Non-Mutually Exclusive Events:

Two events are non-mutually exclusive when they can happen at
the same time.

For example:
Selecting a student who plays cricket and football.

Formula:

P(A or B) = P(A) + P(B) - P(A and B)


IMPORTANT REMINDER:

OR means Addition Rule.

AND means the common part of two events.

Always check whether the events can happen together before
choosing the formula.
"""
```

# PLaying sports

```python

"""
QUESTION 1: PLAYING SPORTS

Question:

In a college, 60% of students play cricket,
40% of students play football,
and 20% of students play both cricket and football.

If a student is selected randomly, find the probability
that the student plays cricket or football.


STEP 1: IDENTIFY THE EVENTS

A = Student plays cricket.

B = Student plays football.

We need to find:

P(A or B)

This means the probability that a student plays cricket
or football or both.


STEP 2: WRITE THE GIVEN PROBABILITIES

P(A) = 0.60

Meaning: Probability that a student plays cricket (60%).

P(B) = 0.40

Meaning: Probability that a student plays football (40%).

P(A and B) = 0.20

Meaning: Probability that a student plays both cricket
and football (20%).


STEP 3: APPLY THE ADDITION RULE

Formula:

P(A or B) = P(A) + P(B) - P(A and B)

Now put the values into the formula:

P(A or B) = 0.60 + 0.40 - 0.20

P(A or B) = 1.00 - 0.20

P(A or B) = 0.80

P(A or B) = 80%


FINAL ANSWER:

The probability that a randomly selected student plays
cricket or football is 80%.


IMPORTANT REMINDER:

We subtract the probability of both sports because those
students were counted twice when adding cricket and football.
"""
```

# python implementation for playing sport

```python

# Identify events
# A = student plays cricket
# B = student plays football

# Given probabilities
p_a = 0.60 # probability of playing cricket
p_b = 0.40 # probability of playing football
p_a_and_b = 0.20 # probability of playing both cricket and football

# apply addition rule
# P(A or B) = P(A) + P(B) - P(A and B)

p_a_or_b = p_a + p_b - p_a_and_b

print("final answer is:", p_a_or_b)
print("percentage:", p_a_or_b * 100, "%")
```

# Drawing a card

```python

"""
QUESTION 2: DRAWING A CARD

Question:

One card is selected randomly from a standard deck of 52 cards.

Find the probability that the selected card is a King
or a Queen.


STEP 1: IDENTIFY THE EVENTS

A = Selected card is a King.

B = Selected card is a Queen.

We need to find:

P(A or B)

This means the probability of selecting a King or a Queen.


STEP 2: WRITE THE GIVEN PROBABILITIES

P(A) = 4/52

Meaning: There are 4 Kings in a deck of 52 cards.

P(B) = 4/52

Meaning: There are 4 Queens in a deck of 52 cards.

P(A and B) = 0

Meaning: One card cannot be both a King and a Queen.

King and Queen are mutually exclusive events.


STEP 3: APPLY THE ADDITION RULE

Formula:

P(A or B) = P(A) + P(B) - P(A and B)

Now put the values into the formula:

P(A or B) = 4/52 + 4/52 - 0

P(A or B) = 8/52

P(A or B) = 2/13

P(A or B) = 15.38% approximately.


FINAL ANSWER:

The probability of selecting a King or a Queen
from a standard deck is approximately 15.38%.


IMPORTANT REMINDER:

A single card cannot be both a King and a Queen.

Therefore, these events are mutually exclusive,
and their common probability is zero.
"""
```

# python implementaion for drawing a card

```python

# Identify events
# A = selected card is a King
# B = selected card is a Queen

# Given probabilities
p_a = 4 / 52 # probability of selecting a King
p_b = 4 / 52 # probability of selecting a Queen
p_a_and_b = 0 # one card cannot be both a King and a Queen

# apply addition rule
# P(A or B) = P(A) + P(B) - P(A and B)

p_a_or_b = p_a + p_b - p_a_and_b

print("final answer is:", p_a_or_b)
print("percentage:", p_a_or_b * 100, "%")
```
