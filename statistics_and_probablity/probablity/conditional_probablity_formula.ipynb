# conditional probablity

# defination and formula

```python

"""
CONDITIONAL PROBABILITY

Definition:
Conditional Probability is used to find the probability
of an event happening when another event has already happened.

In simple words:
It tells us the probability of Event A when we already
know that Event B has occurred.

Formula:

P(A given B) = P(A AND B) / P(B)

Where:
P(A given B) = Probability of A when B has happened.
P(A AND B) = Probability that both A and B happen.
P(B) = Probability that Event B happens.

Important:
P(B) must not be zero.
"""
```

# question 1

```python

"""
QUESTION 1:

In a class of 40 students:
- 20 students play cricket.
- 15 students play both cricket and football.

If a student is selected from those who play cricket,
find the probability that the student also plays football.

SOLUTION:

Identify events:
A = Student plays football.
B = Student plays cricket.

Given:
Total students who play cricket = 20
Students who play both = 15

P(A AND B) = 15/40 is NOT needed here.
Instead, use the number of students who play both
divided by the number who play cricket.

Apply conditional probability:
P(A given B) = 15/20
"""
```

# python impleentation

```python
# Identify events
# A = student plays football
# B = student plays cricket

# Given probabilities
p_a_and_b = 15 / 40
p_b = 20 / 40

# Apply conditional probability
p_a_given_b = p_a_and_b / p_b

print("Final answer is:", p_a_given_b)
print("Percentage:", p_a_given_b * 100, "%")
```

# question 2

```python

"""
QUESTION 2:

A bag contains 5 red balls and 3 blue balls.
Two balls are selected without replacement.

Given that the first ball selected is red,
find the probability that the second ball is also red.

SOLUTION:

Identify events:
A = Second ball is red.
B = First ball is red.

After the first red ball is selected:
Red balls remaining = 4
Total balls remaining = 7

P(A given B) = 4/7
"""
```

# python implemenation

```python
# Identify events
# A = second ball is red
# B = first ball is red

# Given probabilities
p_a_and_b = (5 / 8) * (4 / 7)
p_b = 5 / 8

# Apply conditional probability
p_a_given_b = p_a_and_b / p_b

print("Final answer is:", p_a_given_b)
print("Percentage:", p_a_given_b * 100, "%")
```
