#3rd question

```python
import numpy as np
import matplotlib.pyplot as plt

rng=np.random.default_rng(0)

theory=1 - (1/2)**3

#sample event is is 1 /total events 2
#probablity of at least one head = 1- (1/2)**3
#bcuz we flip coin 3 times thats why prob of getting at least head will become rasied to the power 3

tosses=rng.integers(0,2, size=(100_000,3))

#0 indicates tail and 1 indicates head
got_head=(tosses==1).any(axis=1)
print(got_head)


print("Theory:",round(theory,4))
#it calculates the averages of head we got from tosses
print("simulation we did:", round(got_head.mean(), 4 ))
print("length of got_head:", len(got_head))

running=got_head.cumsum() / np.arange(1,len(got_head)+1)

#thats where matplotlib used to create graphs
plt.plot(running)
plt.axhline(theory,color="red",linestyle="--",label="fair coin tosses")
plt.xlabel("number of games")
plt.ylabel("getting head ")
plt.legend()


plt.show()







```
