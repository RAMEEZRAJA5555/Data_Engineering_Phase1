# Random Variables

# Two types are:

# Discrete random variables and continous random variables

```python

"""
RANDOM VARIABLES

Definition:
A random variable is a variable that represents the numerical
outcome of a random experiment.

It assigns a numerical value to an uncertain outcome.

TYPES OF RANDOM VARIABLES:

1. DISCRETE RANDOM VARIABLE

Definition:
A discrete random variable takes countable and separate
numerical values.

Examples:
- Number of students in a class.
- Number of heads in coin tosses.
- Number of customers in a shop.
- Number of tokens in a prompt.

2. CONTINUOUS RANDOM VARIABLE

Definition:
A continuous random variable can take any numerical
value within a given range.

Examples:
- Height of a student.
- Weight of a person.
- Temperature of a GPU.
- Time taken to load a webpage.

DIFFERENCE:

Discrete = Counting values.
Continuous = Measuring values.

IMPORTANCE IN AI/ML:

Random variables help us represent uncertainty and
understand probability distributions.

Examples:
- Spam detection probability.
- Prediction errors.
- Customer behaviour.
- Temperature predictions.
"""
```

# Python Implementation (Discrete Random Variable)
Question:

A coin is tossed 10 times. Represent Heads as 1 and Tails as 0. Generate 10 random outcomes and count the number of Heads.

Solution:

We use NumPy to generate random outcomes.

```python

import numpy as np

# Generate random outcomes
# 1 = Heads
# 0 = Tails

outcomes = np.random.choice([0, 1], size=10)

# Count the number of Heads
number_of_heads = np.sum(outcomes == 1)

# Display results
print("Coin outcomes:", outcomes)
print("Number of Heads:", number_of_heads)
```

# Python Implementation (Continuous Random Variable)
Question:

A GPU temperature is between 40°C and 80°C. Generate 10 random temperature values and calculate their average.

Solution:

```python

import numpy as np

# Generate random GPU temperatures
# uniform generates equal probablity values within the gievn range,it mean all vales generated between 40 and 80 have same probablity distribution
temperatures = np.random.uniform(40, 80, size=10)

# Calculate average temperature
average_temperature = np.mean(temperatures)

# Display results
print("GPU temperatures:", temperatures)
print("Average temperature:", average_temperature)
```
