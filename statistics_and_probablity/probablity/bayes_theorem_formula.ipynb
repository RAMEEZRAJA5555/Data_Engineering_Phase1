"""
QUESTION 1: FACTORY MACHINES

Question:

A factory has two machines:

- Machine A produces 60% of all products.
- Machine B produces 40% of all products.
- 2% of Machine A's products are defective.
- 5% of Machine B's products are defective.

If a randomly selected product is defective, find the probability that it was made by Machine A.

# formula:
       
                P(A|B) = [P(B|A) × P(A)] / P(B)

STEP 1: IDENTIFY THE EVENTS

A = Product was made by Machine A.
D = Product is defective.
M = Product was made by Machine B.

We need to find:

P(A|D)

This means the probability that the product came from Machine A, given that it is defective.

The product is already known to be defective. We need to find whether it came from Machine A.


STEP 2: WRITE THE GIVEN PROBABILITIES

P(A) = 0.60

Meaning: Probability that the product came from Machine A (60%).

P(M) = 0.40

Meaning: Probability that the product came from Machine B (40%).

P(D|A) = 0.02

Meaning: Probability that a product from Machine A is defective (2%).

P(D|M) = 0.05

Meaning: Probability that a product from Machine B is defective (5%).

IMPORTANT:
The 40% represents products made by Machine B.
It is NOT the total probability of a product being defective.


STEP 3: CALCULATE THE TOTAL PROBABILITY OF A DEFECTIVE PRODUCT

A defective product can come from either Machine A or Machine B.

Formula:

P(D) = (P(D|A) × P(A)) + (P(D|M) × P(M))

Put the values into the formula:

P(D) = (0.02 × 0.60) + (0.05 × 0.40)

P(D) = 0.012 + 0.020

P(D) = 0.032

Therefore, the total probability of getting a defective product is 0.032 or 3.2%.


STEP 4: APPLY BAYES' THEOREM

Formula:

P(A|D) = (P(D|A) × P(A)) / P(D)

Meaning of each value in this formula:

P(A|D):
The probability we need to find. The product came from Machine A, given that it is defective.

P(D|A):
Probability that a product from Machine A is defective (0.02).

P(A):
Probability that the product came from Machine A (0.60).

P(D):
Total probability that a product is defective (0.032).

Now put the values into the formula:

# Formula:
P(A|B) = [P(B|A) × P(A)] / P(B)

P(A|D) = (0.02 × 0.60) / 0.032

P(A|D) = 0.012 / 0.032

P(A|D) = 0.375

P(A|D) = 37.5%


FINAL ANSWER:

If a randomly selected product is defective, the probability that it was made by Machine A is 37.5%.


IMPORTANT REMINDER:

In P(A|D):

D is the condition we already know (the product is defective).

A is what we want to find (the product came from Machine A).
"""


# python implemenation of factory machine example

# Factory Machines — Bayes' Theorem
Question:

A factory has two machines:
- Machine A produces 60% of all products.
- Machine B produces 40% of all products.
- 2% of Machine A's products are defective.
- 5% of Machine B's products are defective.
If a randomly selected product is defective, find the probability that it was made by Machine A

```python

# Identify events
# A(we want to find)=product was made by Machine A
# B(already given in question)=product is defective

# Given probabilities
p_b_given_a = 0.02 # mean 2 percent products of Machine A are defective
p_a = 0.60 # mean initial probability of products made by Machine A

p_b_given_not_a = 0.05 # mean 5 percent products of Machine B are defective
p_not_a = 0.40 # mean initial probability of products made by Machine B

# p_b(total prob)
# P(B) total probability of getting a defective product
# P(B)=P(B|A)*P(A) + P(B|NOT A) * P(NOT A)

p_b = (p_b_given_a * p_a) + (p_b_given_not_a * p_not_a)

# apply bayes theorem
p_a_given_b = (p_b_given_a * p_a) / p_b

print("final answer is:", p_a_given_b)
print("percentage:", p_a_given_b * 100, "%")
```

"""
QUESTION 2: MEDICAL TEST

Question:

A disease affects 1% of a population. A medical test gives a positive result for:

- 90% of people who have the disease.
- 5% of people who do not have the disease.

A person is selected and receives a positive test result. Find the probability that this person actually has the disease.


STEP 1: IDENTIFY THE EVENTS

A = Person has the disease.

B = Test result is positive.

Not A = Person does not have the disease.

We need to find:

P(A|B)

This means the probability that the person has the disease, given that the test result is positive.

The test result is already known to be positive. We need to find whether the person actually has the disease.


STEP 2: WRITE THE GIVEN PROBABILITIES

P(A) = 0.01

Meaning: Probability that a person has the disease (1%).

P(Not A) = 0.99

Meaning: Probability that a person does not have the disease (99%).

P(B|A) = 0.90

Meaning: Probability of a positive test when the person has the disease (90%).

P(B|Not A) = 0.05

Meaning: Probability of a positive test when the person does not have the disease (5%).


STEP 3: CALCULATE THE TOTAL PROBABILITY OF A POSITIVE TEST

A positive test result can happen in two ways:

1. The person has the disease and receives a positive test.
2. The person does not have the disease but still receives a positive test.

Formula:

P(B) = (P(B|A) × P(A)) + (P(B|Not A) × P(Not A))

Put the values into the formula:

P(B) = (0.90 × 0.01) + (0.05 × 0.99)

P(B) = 0.009 + 0.0495

P(B) = 0.0585

Therefore, the total probability of getting a positive test result is 0.0585 or 5.85%.


STEP 4: APPLY BAYES' THEOREM

Formula:

P(A|B) = (P(B|A) × P(A)) / P(B)

Meaning of each value in this formula:

P(A|B):
The probability we need to find. The person has the disease, given that the test result is positive.

P(B|A):
Probability of a positive test when the person has the disease (0.90).

P(A):
Probability that the person has the disease (0.01).

P(B):
Total probability of getting a positive test result (0.0585).

Now put the values into the formula:

P(A|B) = (0.90 × 0.01) / 0.0585

P(A|B) = 0.009 / 0.0585

P(A|B) = 0.1538

P(A|B) = 15.38%


FINAL ANSWER:

Given that a person's test result is positive, the probability that the person actually has the disease is approximately 15.38%.


IMPORTANT REMINDER:

In P(A|B):

B is the condition we already know (the test result is positive).

A is what we want to find (the person has the disease).
"""
