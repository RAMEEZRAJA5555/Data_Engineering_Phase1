# Multiplication rule in probablity

```python

"""
MULTIPLICATION RULE OF PROBABILITY

Definition:
The Multiplication Rule is used to calculate the probability
that two events happen together (Event A AND Event B).

There are two types:

1. Independent Events:
When one event does not affect the probability of another event.

Formula:
P(A AND B) = P(A) * P(B)

2. Dependent Events:
When one event affects the probability of another event.

Formula:
P(A AND B) = P(A) * P(B given A)

Where:
P(A AND B) = Probability that both events happen.
P(A) = Probability of Event A.
P(B) = Probability of Event B.
P(B given A) = Probability of B when A has already happened.

Examples:
Independent: Tossing a coin twice.
Dependent: Selecting two balls without replacement.
"""
```

# terms in multiplication rule
  #1:Independent events
  #2:dependent events

# Question 1:Independent Events

```python

"""
QUESTION 1:

A coin is tossed two times.
Find the probability of getting Heads on both tosses.

SOLUTION:

Identify events:
A = Getting Heads on the first toss.
B = Getting Heads on the second toss.

Given probabilities:
P(A) = 1/2
P(B) = 1/2

Apply multiplication rule:
P(A AND B) = P(A) * P(B)
"""

# Identify events
# A = getting Heads on first toss
# B = getting Heads on second toss

# Given probabilities
p_a = 1 / 2
p_b = 1 / 2

# Apply multiplication rule
p_a_and_b = p_a * p_b

print("Final answer is:", p_a_and_b)
print("Percentage:", p_a_and_b * 100, "%")
```
