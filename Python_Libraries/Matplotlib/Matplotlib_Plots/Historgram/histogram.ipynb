```python
import matplotlib.pyplot as plt

ages = [18, 19, 20, 21, 22, 20, 19, 23, 21, 22,
        24, 20, 19, 21, 25, 22, 23, 20, 21, 19]

plt.hist(ages)

plt.title("Student Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Students")

plt.show()
```
