Continous data

we have range of data for contnious data

```python
"""
## Continuous Data

Continuous data can have any value within a range, including decimal values.

Examples:

* Student height: 165.5 cm
* Student weight: 62.7 kg
* Temperature: 31.6°C
* Time: 2.45 seconds

In a histogram, continuous data is grouped into **ranges (bins)**.
The **X-axis shows the ranges of values**, while the **Y-axis shows how many values fall within each range**.

For example, when checking student heights, a range like **160–165 cm** tells us how many students have heights within that range.


"""
```

```python
import matplotlib.pyplot as plt

#Height is continuous because it can have decimal values.
heights = [
    160.2, 161.5, 162.1, 163.7, 164.3,
    165.2, 165.8, 166.4, 167.1, 168.5,
    169.2, 170.1, 170.8, 171.4, 172.6,
    173.2, 174.5, 175.1, 176.3, 177.0, 178.2
]

plt.hist(heights, bins=5)

plt.title("Distribution of Student Heights")
plt.xlabel("Height")
plt.ylabel("Frequency")

plt.show()

#Here, Matplotlib divides the height range into 5 intervals (bins).
```

Discrete data for histogram


for this we have separte and countable values in x axis

```python
"""
## Discrete Data

Discrete data consists of **separate, countable values**. It usually does not include decimal values.

Examples:

* Number of students: 20
* Number of siblings: 3
* Number of cars: 5
* Number of books: 10

In a histogram, discrete data can also be displayed, but the values are **countable and separate**.

For example, if we check the number of siblings of students, the values could be **0, 1, 2, 3, 4**.

The important difference is:

* **Continuous data → can have any value within a range**
* **Discrete data → separate, countable values**

"""
```

Discrete data — Number of siblings

```python
import matplotlib.pyplot as plt
import seaborn as sns

siblings = [1, 1, 1, 1,1,1,1,1,1
            ]

sns.histplot(siblings,kde=True)


plt.title("Number of Siblings")
plt.xlabel("Number of Siblings")
plt.ylabel("Number of Students")

plt.show()
```
