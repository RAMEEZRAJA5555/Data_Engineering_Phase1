multiple axes mean multiple plots inside one figure,one axes represent one plot area

```python
import matplotlib.pyplot as plt

#Create one Figure with 1 row and 2 columns
#here (1,2) mean figure with 1 row and 2 columns
fig, axes = plt.subplots(1, 2, figsize=(10, 4))


#Check what was created
print(type(fig))
print(type(axes))
print(axes.shape)

#First Axes - Line Plot
axes[0].plot([1, 2, 3, 4], [10, 20, 15, 25])
axes[0].set_title("Line Plot")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Values")

#Second Axes - Bar Plot
axes[1].bar(["A", "B", "C", "D"], [15, 25, 20, 30])
axes[1].set_title("Bar Plot")
axes[1].set_xlabel("Category")
axes[1].set_ylabel("Values")

#Title for the whole Figure
#fig.suptitle adds a title to the whole Figure, not just one Axes.
fig.suptitle("One Figure, Two Axes", fontsize=14, fontweight="bold")

#Adjust spacing
#is used to automatically adjust the spacing between plots and their labels so they don't overlap or get cut off.
plt.tight_layout()

plt.show()
```
