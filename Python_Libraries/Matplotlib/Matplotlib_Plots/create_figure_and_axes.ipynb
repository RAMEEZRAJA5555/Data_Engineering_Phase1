figure is whole canvas or window or we can say blank sheet of paper

axes is plotted area where we drawn plots with x and y labels

we can set figure size too

```python
import matplotlib.pyplot as plt

# Let's make this visible. Create a Figure with one Axes.
fig, ax = plt.subplots(figsize=(10,6))
#We write subplots() because it creates the Figure and Axes for us together.
#Because it is designed to create one or more plotting areas (Axes) inside a Figure.

# Right now it's empty. Let's at least see the structure.
print(type(fig))   # <class 'matplotlib.figure.Figure'>
print(type(ax))    # <class 'matplotlib.axes._subplots.AxesSubplot'>

ax.set_title("This is an empty Axes inside a Figure")
ax.set_xlabel("x-axis label")
ax.set_ylabel("y-axis label")

plt.show()
```
