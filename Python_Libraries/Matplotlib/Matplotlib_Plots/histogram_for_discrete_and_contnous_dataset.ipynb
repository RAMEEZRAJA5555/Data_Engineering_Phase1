```python
import matplotlib.pyplot as plt

heights = [
    160.2, 161.5, 162.1, 163.7, 164.3,
    165.2, 165.8, 166.4, 167.1, 168.5,
    169.2, 170.1, 170.8, 171.4, 172.6,
    173.2, 174.5, 175.1, 176.3, 177.0, 178.2
]

fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(heights, bins=5)

ax.set_title("Distribution of Student Heights")
ax.set_xlabel("Height (cm)")
ax.set_ylabel("Number of Students")

plt.show()
```
