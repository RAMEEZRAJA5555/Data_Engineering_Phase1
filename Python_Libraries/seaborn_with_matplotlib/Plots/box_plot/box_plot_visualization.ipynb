```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data=list(range(1,1000))

plt.boxplot(
    x=data
)

plt.title("box plot visualization")
plt.xlabel("data")
plt.show()

plt.hist(x=data)

plt.show()

```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

arr=list(range(1,10000))

sns.boxplot(x=arr)

plt.title("box plot visualization")
plt.xlabel("data")
plt.show()
```
