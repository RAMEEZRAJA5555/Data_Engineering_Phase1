## bar plot customiztion mean modifying the bar plot:

### Bar Plot Customization

- **estimator** → Calculates the statistical value for each category.
- **hue** → Separates categories using different colors.
- **linewidth** → Controls the thickness of the bar outline.
- **facecolor** → Controls the inside/fill color of the bars.
- **edgecolor** → Controls the outline color of the bars.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "gender": ["Male", "Male", "Male", "Female", "Female", "Female"],
    "math_score": [70, 80, 90, 60, 80, 100],
    "department": ["CS", "SE", "CS", "SE", "CS", "SE"]
})


sns.set_theme(style="whitegrid")

sns.barplot(
    data=data,
    x="gender",
    y="math_score",
    #it separates departments mean it shows each department col as category
    hue="department",
    #calculates the mean
    estimator="mean",
    #controls outline thickness of bars
    linewidth=3,
    #controls bar outline
    edgecolor="yellow"

)

plt.title("Average Math Score by Gender and Department")
plt.xlabel("Gender")
plt.ylabel("Average Math Score")

plt.show()
```
