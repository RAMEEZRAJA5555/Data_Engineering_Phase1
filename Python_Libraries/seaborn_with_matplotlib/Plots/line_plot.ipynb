```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "sales": [120, 135, 150, 145, 170, 180,
              190, 185, 200, 215, 225, 240]
})


sns.lineplot(
    data=data,
    x="month",
    y="sales"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [120, 135, 150, 145, 170, 180]
})

#This sets the overall Seaborn theme and tells Seaborn to use the white-grid style.
"""
# sns.set_theme(style="white")
# sns.set_theme(style="dark")
# sns.set_theme(style="whitegrid")
# sns.set_theme(style="darkgrid")
# sns.set_theme(style="ticks")
"""
sns.set_theme(style="whitegrid")


sns.lineplot(
    data=data,
    x="month",
    y="sales"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```
