```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("2017.csv")

#displays the first five rows by default.
print("First 5 rows")
display(df.head())

#displays the last five rows.
print("Last 5 rows")
display(df.tail())

#returns the number of rows and columns.
print("Shape")
print(df.shape)

#df.columns returns all column names.
#.tolist() converts the column names into a normal Python list.
print("Columns")
print(df.columns.tolist())

#this shows numberof rows,col names,no. of missing vales in each col and each col datatype and memory usage
print("Information")
df.info()

#calculates statistics for numerical columns
print("Statistical summary")
display(df.describe())

#checks whether each value is missing.
print("Missing values")
print(df.isnull().sum())

print("Total missing values")
print(df.isnull().sum().sum())

#identifies rows that duplicate earlier rows.
print("Duplicate rows")
print(df.duplicated().sum())

#df.columns represents the column names.and = assign new names to col
df.columns = [
    "Country",
    "Happiness_Rank",
    "Happiness_Score",
    "Whisker_High",
    "Whisker_Low",
    "Economy",
    "Family",
    "Health",
    "Freedom",
    "Generosity",
    "Trust",
    "Dystopia_Residual"
]

#it will select country col and its values
#.str allows us to perform string operations on its values.
#.strip() removes unnecessary spaces from the beginning and end of each country name.
df["Country"] = df["Country"].str.strip()

df = df.drop_duplicates()

numeric_columns = df.select_dtypes(include="number").columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

print("Cleaned first 5 rows")
display(df.head())

print("Cleaned data types")
print(df.dtypes)

print("Missing values after cleaning")
print(df.isnull().sum())

print("Duplicates after cleaning")
print(df.duplicated().sum())

print("Cleaned shape")
print(df.shape)

print("Cleaned statistical summary")
display(df.describe())

df.to_csv("2017_cleaned.csv", index=False)

print("Cleaned dataset saved successfully")


#sorts countries by happiness score from highest to lowest and selects the first 10.
top_10 = df.sort_values("Happiness_Score", ascending=False).head(10)

#sets the width and height of the chart.
plt.figure(figsize=(10, 5))

#edgecolor defines the borders of each bar.
#color defines the color of the bars.
plt.bar(top_10["Country"], top_10["Happiness_Score"], color="cyan", edgecolor="yellow")

#adds a title to the chart.
plt.title("Top 10 Happiest Countries in 2017")

#adds labels to the x-axis and y-axis.
plt.xlabel("Country")
plt.ylabel("Happiness Score")

#rotates country names for better readability.
plt.xticks(rotation=20)

#adds horizontal grid lines.
#linestyle="--" makes the lines dashed.
#alpha controls transparency.
#transparency means percentage of visibility more alpha more visibility of grid lines
plt.grid(axis="y", linestyle="--", alpha=0.5)

#adjusts the layout to help prevent labels from being cut off.

plt.tight_layout()
#displays the chart.
plt.show()


#sets the size of the chart.
plt.figure(figsize=(8, 5))

#creates a scatter plot to show the relationship between economy and happiness.
#hue adds different colors based on a selected category.
#s sets the size of the points.
#palette sets the colors used in the chart.
sns.scatterplot(data=df, x="Economy", y="Happiness_Score",
                color="purple", s=80)

#adds a title and axis labels.
plt.title("Economy vs Happiness Score")
plt.xlabel("Economy (GDP per Capita)")
plt.ylabel("Happiness Score")

#adds grid lines.
plt.grid(True, linestyle="--", alpha=0.4)

#adjusts the layout.
plt.tight_layout()

#displays the chart.
plt.show()

```
