# Plots limits:

## plot limits:
               Plot limits control which part of the graph is visible.

Imagine your graph is a large window. Your data may cover a big area, but you can choose to show only a specific part of that area.

```python
#xlim → controls what part of X-axis is visible
#ylim → controls what part of Y-axis is visible
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "sales": [120, 135, 150, 145, 170, 180,
              190, 185, 200, 215, 225, 240]
})


sns.set_theme(style="whitegrid")


sns.lineplot(
    data=data,
    x="month",
    y="sales"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xlim(1, 10)    #minimum value for x axes is 1 and max is 10.Only this plot of data will be shown
plt.ylim(100, 220) #it mean minimum value for y axes to show is 100 and maximum value is 220.more then 220 will not be shown.

plt.show()
```
