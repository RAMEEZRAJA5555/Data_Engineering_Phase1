countplot() simply counts the observations.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#it only count how many time male and female appears in data
data = pd.DataFrame({
    "gender": ["Male", "Male", "Female", "Female", "Female", "Male", "Female"]
})

#This sets the overall Seaborn theme and tells Seaborn to use the white-grid style.
sns.set_theme(style="whitegrid")

sns.countplot(data=data, x="gender")

plt.title("Number of Students by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.show()
```

Use of hue:

let's make the Count Plot more useful by adding another category with hue.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = pd.DataFrame({
    "gender": ["Male", "Male", "Female", "Female", "Female", "Male", "Female"],
    "department": ["CS", "SE", "CS", "SE", "CS", "CS", "SE"]
})


sns.countplot(
    data=data,
    x="gender",

    #hue adds another category and separates the count by that category.
    #means we can see the number of CS and SE students separately for each gender.
    hue="department"
)

plt.title("Students by Gender and Department")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.show()
```

Count plot by order:


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = pd.DataFrame({
    "department": [
        "CS", "SE", "CS", "AI", "SE",
        "CS", "AI", "SE", "CS", "AI"
    ]
})


sns.countplot(
    data=data,
    x="department",
    order=["AI", "CS", "SE"]
)

plt.title("Number of Students by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")

plt.show()
```
