Seaborn can understand the DataFrame and automatically calculate things like averages and statistical summaries.

so seaborn is used to do statistical plotting like avergae etc automatically by saying dataframes

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.read_csv("student_performance.csv")


#This sets the overall Seaborn theme and tells Seaborn to use the white-grid style.
"""
# sns.set_theme(style="white")
# sns.set_theme(style="dark")
# sns.set_theme(style="whitegrid")
# sns.set_theme(style="darkgrid")
# sns.set_theme(style="ticks")
"""
sns.set_theme(style="whitegrid")

#Mean (average) — default for barplot
#By default, barplot() uses the mean.
sns.barplot(
    x="gender",
    y="math_score",
    data=data
)

plt.title("Average Math Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Math Score")

plt.show()
```

Use of hue:

let's make the Count Plot more useful by adding another category with hue.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = pd.DataFrame({
    "gender": ["Male", "Male", "Male", "Female", "Female", "Female"],
    "math_score": [70, 80, 90, 60, 80, 100],
    "department": ["CS", "SE", "CS", "SE", "CS", "SE"]
})


sns.barplot(
    data=data,
    x="gender",
    y="math_score",

    ##hue adds another category and separates the count by that category.
    #means we can see the number of CS and SE students separately for each gender.
    hue="department"
)

plt.title("Average Math Score by Gender and Department")
plt.xlabel("Gender")
plt.ylabel("Average Math Score")

plt.show()
```

```python
"""
Bar Plot — Estimator

In Seaborn, the barplot() function uses the mean (average) by default.

We can change the estimator to calculate a different statistic, such as median, minimum, or maximum.

The estimator tells Seaborn which statistical value to calculate for each category.
"""
```

calculate statistical calculation using seaborn Without NumPy:

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = pd.DataFrame({
    "gender": ["Male", "Male", "Male", "Female", "Female", "Female"],
    "math_score": [70, 80, 90, 60, 80, 100]
})


# Mean
sns.barplot(data=data, x="gender", y="math_score")
plt.title("Mean Math Score")
plt.show()


# Median
sns.barplot(data=data, x="gender", y="math_score", estimator="median")
plt.title("Median Math Score")
plt.show()


# Minimum
sns.barplot(data=data, x="gender", y="math_score", estimator="min")
plt.title("Minimum Math Score")
plt.show()


# Maximum
sns.barplot(data=data, x="gender", y="math_score", estimator="max")
plt.title("Maximum Math Score")
plt.show()
```

With NumPy:

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np


data = pd.DataFrame({
    "gender": ["Male", "Male", "Male", "Female", "Female", "Female"],
    "math_score": [70, 80, 90, 60, 80, 100]
})


# Mean
sns.barplot(data=data, x="gender", y="math_score")
plt.title("Mean Math Score")
plt.show()


# Median
sns.barplot(data=data, x="gender", y="math_score", estimator=np.median)
plt.title("Median Math Score")
plt.show()


# Minimum
sns.barplot(data=data, x="gender", y="math_score", estimator=np.min)
plt.title("Minimum Math Score")
plt.show()


# Maximum
sns.barplot(data=data, x="gender", y="math_score", estimator=np.max)
plt.title("Maximum Math Score")
plt.show()
```
