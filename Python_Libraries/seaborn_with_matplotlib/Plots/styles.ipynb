## Styles:

Seaborn style controls the overall visual appearance of a plot, such as its background, grid, and axis look.
You can write:

sns.set_theme(style="whitegrid")

Meaning: Apply the whitegrid style to the plot.

### Seaborn Styles

- **white** → Clean white background
- **dark** → Dark background
- **whitegrid** → White background with grid
- **darkgrid** → Dark background with grid
- **ticks** → Clean background with visible axis ticks

### syntax to write style:

sns.set_theme(style="white")

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "sales": [120, 135, 150, 145, 170, 180]
})

#This sets the overall Seaborn theme and tells Seaborn to use the white-grid style.
"""
# sns.set_theme(style="white")
# sns.set_theme(style="dark")
# sns.set_theme(style="whitegrid")
# sns.set_theme(style="darkgrid")
# sns.set_theme(style="ticks")
"""
sns.set_theme(style="whitegrid")


sns.lineplot(
    data=data,
    x="month",
    y="sales"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

## for scatter plot:

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = pd.DataFrame({
    "age": [18, 20, 22, 25, 28, 30, 32],
    "salary": [30000, 35000, 40000, 45000, 50000, 55000, 60000]
})

#This sets the overall Seaborn theme and tells Seaborn to use the white-grid style.
sns.set_theme(style="whitegrid")

sns.scatterplot(data=data, x="age", y="salary")

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```
