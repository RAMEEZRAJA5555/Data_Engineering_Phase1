## heatmap concepts:

#A heatmap represents numerical values using colors. Different colors or color intensities help us compare values quickly.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = pd.DataFrame({
    "Python": [85, 70, 90, 60],
    "Database": [75, 80, 95, 65],
    "Data Analysis": [90, 85, 80, 70]
}, index=["Ali", "Ahmed", "Sara", "Ayesha"])
#index= assigns student names to the rows.

sns.set_theme(style="white")
# annot=true displays the actual numbers inside the colored cells.
sns.heatmap(data, annot=True)

plt.title("Students' Marks Heatmap")

plt.show()
```
