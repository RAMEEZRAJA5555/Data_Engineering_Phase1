```python
import pandas as pd

df=pd.read_csv("employe salary.csv")

print(df.info())
print(df.duplicated())

#it will give rows where salaray is more than 100000
print(df[(df["Monthly_Salary"] > 100000)])

#it willl give rows where both condition are true
print(df[(df["Monthly_Salary"] > 100000) & (df["Age"] > 20)])

df["Name"]=df["Name"].str.upper()
print(df["Name"])

#increases each employee salary by 10 percent
df["Grand_Salary"]=df["Monthly_Salary"] * 1.10
print(df["Grand_Salary"])

#Find the average monthly salary of employees in each department.
print(df.groupby("Department")["Monthly_Salary"].mean())

#Find the total monthly salary paid by each department
print(df.groupby("Department")["Monthly_Salary"].sum())

#Find how many employees are in each department.
print(df.groupby("Department")["EmployeeID"].count())

#This puts the employee with the lowest salary first and highest last.
print(df.sort_values("Monthly_Salary"))

#This puts the highest-paid employee first.
print(df.sort_values("Monthly_Salary", ascending=False))

#Sort by Department, then Salary
#Department → A to Z
#Within the same department → salary low to high
print(df.sort_values(["Department", "Monthly_Salary"]))

#Sort by Department A-Z and Salary high-low
print(df.sort_values(["Department", "Monthly_Salary"], ascending=[True, False]))








```
