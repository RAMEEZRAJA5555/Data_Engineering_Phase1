```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Hina", "Usman", "Ayesha"],
    "Age": [20, 22, 19, 24, 26, 21],
    "Salary": [50000, 60000, 45000, 70000, 55000, 65000],
    "City": ["Multan", "Lahore", "Karachi", "Lahore", "Multan", "Islamabad"]
}

df=pd.DataFrame(data)
print(df)

#Add a new column with the same value
#it will make new col and will assign 5000 to each row of dataframe
df["Bonus"] = 5000

print(df)
print(df["Bonus"])

#Add a calculated column
#it will make nw col total salary and calculate two col
df["total Salary"]=df["Salary"] + df["Bonus"]
print(df["total Salary"])
print(df)

#Modify an existing column
#it will increase each salary by 10 percent
df["Salary"]=df["Salary"] *1.10
print(df["Salary"])
print(df)


#Modify a string column
#it mean write col name with all upper case letters
df["Name"] = df["Name"].str.upper()
print(df)







```
