```python
#filtering means selecting only the rows (or data) that satisfy a specific condition.
#filtering is use when we want to see which emplyee have max salay,or which student have max marks etc,bcuz these only gets specific rows
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Hina", "Usman", "Ayesha"],
    "Age": [20, 22, 19, 24, 26, 21],
    "Salary": [50000, 60000, 45000, 70000, 55000, 65000],
    "City": ["Multan", "Lahore", "Karachi", "Lahore", "Multan", "Islamabad"]
}

df=pd.DataFrame(data)

print(df)

#Filter employees whose age is greater than 20
print(df[df["Age"] > 20 ])

#Filter employees whose salary is greater than 55000
print(df[df["Salary"] > 55000 ])

#Filter employees from Lahore
print(df[df["City"] == "Lahore"])

#filtering using multiple conditions
#AND condition
print(df[(df["Age"] > 20) & (df["Salary"] > 55000)])

#OR condition
print(df[(df["City"] == "Lahore") | (df["City"] == "Multan")])

#Filter using isin()
#Check whether each value in the City column is present in this list.
print(df[df["City"].isin(["Lahore", "Multan"])])  #this line done the sae work as above one,
```
