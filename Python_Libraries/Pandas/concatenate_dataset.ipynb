pd.concat()


```python
#pd.concat:
#pd.concat() is used to combine two or more Pandas DataFrames.



#There are two main ways:

#A. Concatenate rows — one DataFrame below another
#B. Concatenate side-by-side(colums) 
```

```python
#concaenate for rows is used when colum names are same for both dataframes

#concatenate side by side is used when dataframes have differnt colums
```

Concatenate rows


```python
#concatenate rows --mean one dataframe belwo another

import pandas as pd

df1=pd.DataFrame({
    "Name":["Ali","Salman","Ahmed","Akash"],
    "Age":[20,21,22,23]

})


df2=pd.DataFrame({
    "Name":["Haider","Muzammil","Yousaf","islam"],
    "Age":[25,27,29,27]

})

results=pd.concat([df1,df2] , ignore_index=True)
print(results)
```

Concatenate side by side


```python
import pandas as pd

df1=pd.DataFrame({
    "Name":["Ali","Salman","Ahmed","Akash"],
    "Father":["shiekh","nawaz","naeem","aslam"]

})


df2=pd.DataFrame({
    "grandfather":["Haider","Muzammil","Yousaf","islam"],
    "grandmother":["sakeena","manzoora","bibi","mai"]

})

results=pd.concat([df1,df2],axis=1)
print(results)
```

```python
#lets create 3 related files for concatenation

import pandas as pd

df1=pd.read_csv("students_2024.csv")
df2=pd.read_csv("students_2025.csv")
df3=pd.read_csv("students_2026.csv")

results=pd.concat([df1,df2,df3], ignore_index=True)
print(results)
```
