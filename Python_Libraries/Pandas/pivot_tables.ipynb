A pivot table takes detailed/raw data and creates a summary table by grouping the data according to categories.

Pivot Tables

A pivot table is used to summarize data by categories and arrange the result into a useful table.

You can think of it as:
                     Take my detailed dataset and create a summary table showing how values compare across categories

```python
"""
It is especially useful when you want to analyze things like:

average score by department
total sales by product and month
average salary by city and department
IPL runs by team and season
number of employees by department and gender
"""
```

Basic pivot table

```python
import pandas as pd


df = pd.DataFrame({
    "Department": ["CS", "CS", "SE", "SE", "AI", "AI"],
    "Gender": ["M", "F", "M", "F", "M", "F"],
    "Score": [80, 90, 75, 85, 95, 88]
})

#suppose we want:
#Average score for each department, separated by gender.

#we can use:
summary=pd.pivot_table(
    df,                   #The dataframe we are analyzing
    values="Score",       #The numerical column we want to calculate something on.
    index="Department",   #The category that becomes the rows.
    columns="Gender",     #The category that becomes the columns.
    aggfunc="mean"        #What calculation should be performed?
)

print(df)
print(summary)

#after that:
#So the pivot table has transformed the original detailed data into a summary view.
```

More concepts for Pivot_table


```python
import pandas as pd

df = pd.DataFrame({
    "Student": [
        "Ali", "Sara", "Ahmed",
        "Hina", "Usman", "Ayesha",
        "Hamza", "Zainab", "Bilal",
        "Omar", "Laiba"
    ],
    "Department": [
        "CS", "CS", "CS",
        "SE", "SE", "SE",
        "AI", "AI", "AI",
        "CS", "AI"
    ],
    "Category": [
        "ML", "Analytics", "ML",
        "ML", "Analytics", "ML",
        "ML", "Analytics", "ML",
        "Analytics", "ML"
    ],
    "Year": [
        2, 2, 3,
        2, 3, 3,
        2, 3, 3,
        3, 2
    ],
    "Score": [
        85, 78, 90,
        75, 88, 82,
        95, 89, 91,
        84, 94
    ],
    "Adjusted_Score": [
        76.5, 70.2, 81.0,
        67.5, 79.2, 73.8,
        85.5, 80.1, 81.9,
        75.6, 84.6
    ]
})

print(df)

#What is the average Adjusted_Score for each Department and Category?
pivot_mean = df.pivot_table(
    values="Adjusted_Score",
    index="Department",
    columns="Category",
    aggfunc="mean"
)

print(pivot_mean)

#What is the total adjusted score for each Department × Category?
pivot_sum = df.pivot_table(
    values="Adjusted_Score",
    index="Department",
    columns="Category",
    aggfunc="sum"
)

print(pivot_sum)

#How many students are in each Department × Category?
pivot_count = df.pivot_table(
    values="Adjusted_Score",
    index="Department",
    columns="Category",
    aggfunc="count"
)

print(pivot_count)
```

all concepts for pivot table 

```python
import pandas as pd

# Load data
df = pd.read_csv("students_pivot.csv")

# Complete pivot table
pivot = df.pivot_table(
    values=["Score", "Adjusted_Score"],
    index="Department",
    columns="Category",
    aggfunc=["mean", "sum", "count", "max", "min"],
    fill_value=0,
    margins=True,     #(Also give me the overall summary.")It adds an extra total/overall row and column to your pivot table.
    margins_name="Total"  #The All column gives the average for each department without separating by category.
).round(2)


#main dataset
print(df)
#summary of data
print(pivot)
```
