````python
"""
## Pandas `merge()`

`merge()` is used to combine two DataFrames by matching rows based on one or more common columns (keys).

It is useful when two datasets contain different information about the same entities.

For example:

* `students` → contains student information
* `scores` → contains student scores
* `Student_ID` → common key used to match them

### Why do we use `merge()`?

We use `merge()` when we want to bring related information from different DataFrames into one DataFrame.

Example:

`students` + `scores` → one DataFrame containing both student information and scores.

### Basic Syntax

```python
pd.merge(left_dataframe, right_dataframe, on="column", how="merge_type")
```

* `on` → common column used to match rows
* `how` → specifies which rows should be kept

### Types of Merge

* `inner` → keeps only matching rows from both DataFrames
* `left` → keeps all rows from the left DataFrame and matching rows from the right
* `right` → keeps all rows from the right DataFrame and matching rows from the left
* `outer` → keeps all rows from both DataFrames

### Important Difference

`merge()` matches related data using a key.

`concat()` mainly combines DataFrames by stacking rows or columns.

In simple words:

`concat()` → combine datasets

`merge()` → match and combine related datasets

"""
````

````python
import pandas as pd

import pandas as pd

students = pd.read_csv("student_info.csv")
scores = pd.read_csv("scores_data.csv")

print(students)
print(scores)

#Inner merge(default)
#inner keeps only the matching records from both DataFrames.

result1 = pd.merge(   
    students,
    scores,
    on="Student_ID",
    how="inner"
)

print(result1)

#Left Merge
#Keep everything from the left DataFrame (students) and bring matching information from scores.

result2 = pd.merge(
    students,
    scores,
    on="Student_ID",
    how="left"
)

print(result2)

#right merge

result3 = pd.merge(
    students,
    scores,
    on="Student_ID",
    how="right"
)

print(result3)


#ouer merge
#This is also called a full outer merge.

result4 = pd.merge(
    students,
    scores,
    on="Student_ID",
    how="outer"
)

print(result4)
````
