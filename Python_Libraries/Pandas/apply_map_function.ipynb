```python
#apply() and map() are useful when adding or modifying columns based on a rule or transformation.

#apply() lets you run a function on each value in a column.

#apply() lets you run a function on each value in a column.
#map() → replace/transform values according to a mapping.
```

```python

#apply() is used to apply rule or function on the dataframe column
#And return new value and store in new column
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Hina"],
    "Age": [25, 30, 35, 28],
    "Salary": [50000, 80000, 120000, 60000]
})

print(df)

def Salary_Level(x):
    if x >=100000:
        return "High"
    else:
        return "Low"


#Pandas automatically takes each salary value from "Salary" and gives it to the function:
df["Salary_Level"]=df["Salary"].apply(Salary_Level)
print(df)


#when we want to apply on rows

def check_employe(row):  #here row contain the whole row so we can access.
    if (row["Age"] > 20 ) & (row["Salary"] >= 50000) :
        return "Senior Highr Salary"
    else:
        "still Junior"

df["Category"]=df.apply(check_employe , axis=1)
print(df["Category"])
print(df)
```

```python
#map() is commonly used on one column to convert/map each value to another value.


import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Hina"],
    "Age": [25, 30, 35, 28],
    "Salary": [50000, 80000, 120000, 60000],
    "Department":["Artificail Intelligence","Computer Science","Marketing","Data Science"]
})

print(df)

department_code={
    "Artificail Intelligence":"AI",
    "Computer Science":"CS",
    "Marketing":"MKT",
    "Data Science":"DS"
}

df["Department_code"]=df["Department"].map(department_code)
print(df["Department_code"])
print(df)
```
