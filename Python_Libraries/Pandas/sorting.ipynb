```python
#Sorting means arranging rows according to the values in a column.
#sorting is used to arrange your data in a specific order.
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Hina", "Usman", "Ayesha"],
    "Age": [20, 22, 19, 24, 26, 21],
    "Salary": [50000, 60000, 45000, 70000, 55000, 65000],
    "City": ["Multan", "Lahore", "Karachi", "Lahore", "Multan", "Islamabad"]
}

df = pd.DataFrame(data)
print(df)

#Sort salary from smallest to largest
#by default order is ascending
print(df.sort_values("Salary"))


#Sort salary from largest to smallest
print(df.sort_values("Salary",ascending=False))

#Sort age from smallest to largest
print(df.sort_values("Age"))

#Sort by multiple columns
#it will sort "city" first alphabatically A--->Z
#Then sort by Salary → for employees who have the same city, sort their salaries from low → high.
print(df.sort_values(["City", "Salary"]))

#This means City A→Z, but Salary high→low within each city.
df.sort_values(["City", "Salary"], ascending=[True, False])


```
