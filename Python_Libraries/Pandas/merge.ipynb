```python
#pd.merge() is used to combine two DataFrames together using a common column.

#Think of it like joining two tables in SQL.
```

```python
#Merging is combining two DataFrames using a common column that exists in both DataFrames.

import pandas as pd

students = pd.read_csv("students.csv")
scores = pd.read_csv("scores.csv")

merged = pd.merge(students, scores, on="student_id")

print(merged)
```
