```python
#groupby() is used to group rows that have the same value, so you can then perform calculations on each group.
#Here we want to answer questions such as:

#What is the average salary in each city?
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Hina", "Usman", "Ayesha", "Bilal", "Zara"],
    "Department": ["IT", "HR", "IT", "HR", "IT", "Finance", "Finance", "IT"],
    "City": ["Multan", "Lahore", "Karachi", "Lahore", "Multan", "Islamabad", "Multan", "Lahore"],
    "Salary": [50000, 60000, 45000, 70000, 55000, 65000, 48000, 75000]
}

df = pd.DataFrame(data)
print(df)

print(df.info())

#Average salary by city
#here in city col there are many same values so we will create group for this and we will so calculation on each group
#if values are not same,then groupby will still work,then each value will become their own group
print(df.groupby("City")["Salary"].mean())

#Total salary by city
print(df.groupby("City")["Salary"].sum())

#Number of employees in each city
print(df.groupby("City")["Name"].count())

#Minimum salary by city
print(df.groupby("City")["Salary"].min())

#Maximum salary by city
print(df.groupby("City")["Salary"].min())

#Average salary by department
print(df.groupby("Department")["Salary"].mean())




```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Hina", "Usman", "Ayesha"],
    "Department": ["IT", "IT", "HR", "HR", "Finance", "Finance"],
    "Salary": [80000, 100000, 60000, 70000, 90000, 110000]
})

print(df)

#it means For each department, calculate the average salary.
result1=df.groupby("Department")["Salary"].agg("mean")
print(result1)

#df.groupby("Department")["Salary"].mean()


#.agg() with multiple functions

result2 = df.groupby("Department")["Salary"].agg(
    ["mean", "max", "min"]
)

print(result2)

"""
You can also give the results your own names

result = df.groupby("Department").agg(
    average_salary=("Salary", "mean"),
    highest_salary=("Salary", "max"),
    lowest_salary=("Salary", "min")
)

print(result)
"""

#.transform()

#We want to add a new column showing the average salary of that employee's department.

df["Department_Avg"] = (
    df.groupby("Department")["Salary"].transform("mean")
)
print(df["Department_Avg"])

print(df)



```
